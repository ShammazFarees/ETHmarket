```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.stattools import adfuller
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_squared_error, mean_absolute_percentage_error
from math import sqrt
import yfinance as yf
from datetime import datetime, timedelta
```

```python
## 1. Data Collection & Preparation

# Define the ticker symbol and date range
ticker = "ETH-USD"
start_date = "2020-01-01"
end_date = datetime.now().strftime("%Y-%m-%d")

# Download historical data from Yahoo Finance
print(f"Downloading Ethereum (ETH/USDT) data from {start_date} to {end_date}...")
eth_data = yf.download(ticker, start=start_date, end=end_date)

# Check data
print("\nFirst 5 rows of the dataset:")
print(eth_data.head())

# Data Cleaning and Preparation
print("\nData cleaning and preprocessing...")

# Check for missing values
print("\nMissing values before cleaning:")
print(eth_data.isnull().sum())
```

```python

## 3. Stationarity Testing

# Function to perform ADF test and print results
def adf_test(timeseries):
    print("Results of Augmented Dickey-Fuller Test:")
    dftest = adfuller(timeseries, autolag='AIC')
    dfoutput = pd.Series(dftest[0:4], index=['Test Statistic', 'p-value', '#Lags Used', 'Number of Observations Used'])
    for key, value in dftest[4].items():
        dfoutput[f'Critical Value ({key})'] = value
    print(dfoutput)
    
    # Plot the time series being tested
    plt.figure(figsize=(12, 5))
    plt.plot(timeseries, label='Original Series')
    plt.title('Stationarity Test Visualization')
    plt.legend()
    plt.grid(True)
    plt.show()
    
    return dftest[1]  # return p-value
```

```python
# Test stationarity of closing prices
print("\nTesting stationarity of raw closing prices...")
p_value_raw = adf_test(eth_data['Close'].dropna())

# S
print("\nTesting stationarity of first differences...")
p_value_diff = adf_test(eth_data['Close'].diff().dropna())
```

```python
# Determine if differencing is needed
if p_value_raw > 0.05:
    print("\nConclusion: The raw time series is non-stationary (p-value > 0.05). Differencing is required.")
    if p_value_diff <= 0.05:
        print("After first differencing, the series becomes stationary (p-value ≤ 0.05).")
        d = 1  # first difference is sufficient
    else:
        print("Even after first differencing, the series remains non-stationary. Further differencing may be needed.")
        # In practice, we rarely need more than second differencing for financial time series
        d = 2
else:
    print("\nConclusion: The raw time series is stationary (p-value ≤ 0.05). No differencing needed.")
    d = 0
    
```

```python
## 4. ARIMA Model Development

# Plot ACF and PACF to determine ARIMA parameters
plt.figure(figsize=(12, 8))

# ACF plot
plt.subplot(2, 1, 1)
plot_acf(eth_data['Close'].diff().dropna(), lags=40, ax=plt.gca(), title='Autocorrelation Function (ACF)')
plt.grid(True)
```

```python
# PACF plot
plt.subplot(2, 1, 2)
plot_pacf(eth_data['Close'].diff().dropna(), lags=40, ax=plt.gca(), title='Partial Autocorrelation Function (PACF)')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Interpret ACF/PACF plots to determine p and q
print("\nInterpreting ACF/PACF plots:")
print("- ACF shows a gradual decay, suggesting an AR component may be appropriate.")
print("- PACF has significant spikes at lags 1 and possibly 2, suggesting p=1 or p=2.")
print("- ACF has a significant spike at lag 1, suggesting q=1.")
print("- We'll test multiple combinations around these values.")
```

```python
# Split data into train and test sets (80-20 split)
train_size = int(len(eth_data) * 0.8)
train, test = eth_data['Close'][:train_size], eth_data['Close'][train_size:]
```

```python
# Function to evaluate ARIMA models
def evaluate_arima_model(train_data, test_data, order):
    # Prepare model
    model = ARIMA(train_data, order=order)
    model_fit = model.fit()
    
    # Forecast
    forecast = model_fit.forecast(steps=len(test_data))
    
    # Calculate error metrics
    rmse = sqrt(mean_squared_error(test_data, forecast))
    mape = mean_absolute_percentage_error(test_data, forecast) * 100
    
    return rmse, mape, model_fit
```

```python
# Test multiple ARIMA configurations
print("\nTesting different ARIMA configurations...")

# Candidate models to test
models = [
    (1, d, 1),  # Based on ACF/PACF interpretation
    (2, d, 1),  # Trying higher AR term
    (1, d, 2),  # Trying higher MA term
    (2, d, 2),  # Combination
    (0, d, 1),  # Simple MA model
    (1, d, 0)   # Simple AR model
]

best_rmse = float('inf')
best_mape = float('inf')
best_order = None
best_model = None

for order in models:
    try:
        rmse, mape, model_fit = evaluate_arima_model(train, test, order)
        print(f"ARIMA{order} - RMSE: {rmse:.2f}, MAPE: {mape:.2f}%")
        
        if rmse < best_rmse:
            best_rmse = rmse
            best_mape = mape
            best_order = order
            best_model = model_fit
    except:
        continue
print(f"\nBest model: ARIMA{best_order} with RMSE: {best_rmse:.2f}, MAPE: {best_mape:.2f}%")

# Fit the best model on the entire dataset
final_model = ARIMA(eth_data['Close'], order=best_order)
final_model_fit = final_model.fit()

# Print model summary
print("\nFinal Model Summary:")
print(final_model_fit.summary())

# Plot diagnostics
final_model_fit.plot_diagnostics(figsize=(12, 8))
plt.tight_layout()
plt.show()
```

```python
## 5. Model Evaluation

# Make in-sample predictions
predictions = final_model_fit.predict(start=train_size, end=len(eth_data)-1, typ='levels')

# Calculate final evaluation metrics
final_rmse = sqrt(mean_squared_error(test, predictions))
final_mape = mean_absolute_percentage_error(test, predictions) * 100

print("\nFinal Model Evaluation on Test Set:")
print(f"RMSE: {final_rmse:.2f}")
print(f"MAPE: {final_mape:.2f}%")
```

```python
# Plot actual vs predicted values
plt.figure(figsize=(12, 6))
plt.plot(test.index, test, label='Actual Price', color='blue')
plt.plot(test.index, predictions, label='Predicted Price', color='orange', linestyle='--')
plt.title('Actual vs Predicted Ethereum Prices')
plt.ylabel('Price (USDT)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Plot residuals
residuals = test - predictions
plt.figure(figsize=(12, 6))
plt.plot(residuals)
plt.title('Residuals of ARIMA Model')
plt.ylabel('Residual')
plt.grid(True)
plt.show()
```

```python
## 6. Forecasting & Visualization

# Forecast the next 30 days
forecast_steps = 30
forecast = final_model_fit.get_forecast(steps=forecast_steps)
forecast_mean = forecast.predicted_mean
forecast_ci = forecast.conf_int()
# Create date range for forecast period
last_date = eth_data.index[-1]
forecast_dates = pd.date_range(start=last_date + timedelta(days=1), periods=forecast_steps)

# Plot historical data and forecast
plt.figure(figsize=(14, 7))
```

```python
# Plot historical data (last 180 days for clarity)
plt.plot(eth_data.index[-180:], eth_data['Close'][-180:], label='Historical Price', color='blue')

# Plot forecast
plt.plot(forecast_dates, forecast_mean, label='Forecast', color='red')
plt.fill_between(forecast_dates, 
                 forecast_ci.iloc[:, 0], 
                 forecast_ci.iloc[:, 1], 
                 color='pink', alpha=0.3, label='95% Confidence Interval')

plt.title(f'Ethereum (ETH/USDT) Price Forecast - Next {forecast_steps} Days')
plt.ylabel('Price (USDT)')
plt.xlabel('Date')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Create a DataFrame with the forecast results
forecast_df = pd.DataFrame({
    'Date': forecast_dates,
    'Forecasted_Price': forecast_mean,
    'CI_Lower': forecast_ci.iloc[:, 0],
    'CI_Upper': forecast_ci.iloc[:, 1]
})

print("\nForecasted Prices for the Next 30 Days:")
print(forecast_df.head(10))  # Show first 10 days
```
